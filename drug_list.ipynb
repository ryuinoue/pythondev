```python
import csv

# 薬のサンプルデータを作成（2次元リスト）
drug_data = [
    ["薬品名", "一般的な用量", "注意点"],
    ["アセトアミノフェン坐剤100mg", "10-15mg/kg/回", "空腹時を避けるのが望ましい"],
    ["カルボシステインシロップ5%", "体重kgあたり1mL", "よく振ってから服用"],
    ["アスベリン散10%", "1歳以上3歳未満: 0.2g/日", "眠気が出ることがある"]
]

# "drug_list.csv" という名前でファイルに書き込む
# encoding="utf-8-sig" にするとExcelで開いた時の文字化けを防ぎやすい
with open("drug_list.csv", "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(drug_data)

print("drug_list.csv ファイルを作成しました。左側のファイル一覧から確認してみてください。")

# 今しがた保存したCSVファイルを読み込んでみる
with open("drug_list.csv", "r", encoding="utf-8-sig") as f:
    reader = csv.reader(f)
    # 1行ずつ読み込んで表示する
    for row in reader:
        print(row)
```

```python
import re

text = "アセトアミノフェン 10-15mg/kg/回"
pattern = r"(\d+)-(\d+)"

# findallはタプルのリスト[('10','15')]を返す
result_of_tuples = re.findall(pattern,text)

# もし結果が見つかっていれば、
# リストの最初の要素であるタプル（’10’,'15'）を取り出し、
# それをリストに追加する

if result_of_tuples:
    final_list = list(result_of_tuples[0])
else:
    final_list = [] # マッチしなかった場合は空のリスト

print(final_list)
```

```python
import re

#チェックしたいメールアドレスのリスト
email_to_check = [
    "test@example.com",        # OK
    "ryuinoue@ncc.go.jp",     # OK
    "user123@sub.domain.org",  # 今回の簡易版ではNG
    "test@examplecom",         # NG（ドット抜け）
    "test.example.com",        # NG(@抜け)
    "@example.com"            # ユーザー抜け
]

# シンプルなメールアドレスパターン
# ^\w+@\w+\.\w+$ の意味：
# ^       : 文字列の先頭
# \w+     : 1文字以上の英数字_
# @       : @
# \w+     : 1文字以上の英数字_
# \.      : ドット
# \w+     : 1文字以上の英数字_
# $       : 文字列の末尾

pattern = r"^\w+@\w+\.\w+$"

print("---メールアドレス　チェック開始---")

for email in email_to_check:
    if re.match(pattern,email):
        print(f"'{email}' は、有効な形式です。")
    else:
        print(f"'{email}' は、無効な形式です。")
        
```

```python
import re

#チェックしたいメールアドレスのリスト
email_to_check = [
    "test@example.com",        # OK
    "ryuinoue@ncc.go.jp",     # OK
    "user123@sub.domain.org",  # 今回の簡易版ではNG
    "test@examplecom",         # NG（ドット抜け）
    "test.example.com",        # NG(@抜け)
    "@example.com"            # ユーザー抜け
]

# シンプルなメールアドレスパターン
# ^\w+@\w+\.\w+$ の意味：
# ^       : 文字列の先頭
# \w+     : 1文字以上の英数字_
# @       : @
# \w+     : 1文字以上の英数字_
# \.      : ドット
# \w+     : 1文字以上の英数字_
# $       : 文字列の末尾

pattern = r"^\w+@(\w+\.)+\w+$"

print("---メールアドレス　チェック開始---")

for email in email_to_check:
    if re.match(pattern,email):
        print(f"'{email}' は、有効な形式です。")
    else:
        print(f"'{email}' は、無効な形式です。")
```

```python
# --- ステップ１：必要なライブラリーインポートとデータ準備　--
import numpy as np
import matplotlib.pyplot as plt

# 課題で与えられているデータ（英語と数学の点数）

x = np.array([62,95,83,100,72,72,55,67,75])    # English
y = np.array([62,80,85,58,76,52,84,67,71])     # math

# 近似直線の計算（最小二乗法）

p = np.polyfit(x,y,1)
f = np.polyld(p)


# --- ステップ２ ＆ ３:グラフの描画と統合 ---

# 散布図を描画(点:青色,ラベ:'score')
plt.scatter(x,y,color="blue",label="score")

# 近似線を重ねて描画(線：黄色、ラベル：’lsm’)
plt.plot(x,f(x),color="red",label="lsm")

# 軸のラベルを設定
plt.xlabel("english")
plt.ylabel("math")

# 凡例表示(左上に配置)
plt.legend(loc="upper left")

#グラフ表示
plt.show
```
