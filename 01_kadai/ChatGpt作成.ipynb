```python
def main():
    total = 0
    values = []

    while True:
        user_input = input("値を入力してください：")
        # 未入力の場合、ループ終了
        if user_input == "":
            break
        try:
            num = float(user_input)
            values.append(num)
            total += num
        except ValueError:
            print("有効な数値を入力してください。")
    
    # 入力された値を表示
    print("\n入力された値:", values)
    # 合計値を表示
    print("合計値:", total)

if __name__ == "__main__":
    main()
    
```

```python
def main():
    items = []
    total = 0

    while True:
        product_name = input("商品名を入力してください：")
        if product_name == "":
            break
        price_input = input("値段を入力してください：")
        if price_input == "":
            # 値段未入力はスキップまたはエラーにするか検討
            print("値段が未入力のためスキップします。")
            continue
        try:
            price = int(price_input)
        except ValueError:
            print("有効な整数値を入力してください。")
            continue
        
        items.append((product_name, price))
        total += price

    # 商品名でソート
    sorted_items = sorted(items, key=lambda x: x[0])

    # 結果表示
    print("\n商品名でソートした一覧：")
    for name, price in sorted_items:
        print(f"{name}: {price}")
    print("合計金額:", total)

if __name__ == "__main__":
    main()
```
